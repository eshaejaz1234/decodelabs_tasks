```python
preferences = {
    "Movies": ["Action Movie", "Comedy Movie", "Sci-Fi Movie"],
    "Books": ["Python Book", "AI Book", "Data Science Book"],
    "Music": ["Pop Music", "Classical Music", "Rock Music"]
}

print("Choose your interest:")
print("1. Movies")
print("2. Books")
print("3. Music")

choice = input("Enter your choice: ")

if choice == "1":
    print("Recommended:", preferences["Movies"])
elif choice == "2":
    print("Recommended:", preferences["Books"])
elif choice == "3":
    print("Recommended:", preferences["Music"])
else:
    print("Invalid Choice")
```
